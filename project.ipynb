```python
# ==================================================
# Harry Potter Text Generation with LSTM
# Project 1446 - Complete Implementation
# ==================================================

# Import all necessary libraries
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

import nltk
from nltk.tokenize import word_tokenize
import re
import string
import os
import warnings
warnings.filterwarnings('ignore')

# Download NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')

print("✓ All libraries imported successfully!")
print(f"TensorFlow Version: {tf.__version__}")

# ==================================================
# 1. DATASET SELECTION & LOADING
# ==================================================
print("\n" + "="*50)
print("1. DATASET SELECTION")
print("="*50)
print("Selected: Harry Potter Fan Fiction Dataset")

# Load Harry Potter dataset
# Note: You'll need to upload the dataset file in Colab
# For this example, I'll show how to load from uploaded file

from google.colab import files
import io

print("\nPlease upload your Harry Potter dataset (CSV or TXT file)")
print("If you don't have the file, we'll use sample text for demonstration")

# Sample Harry Potter text (fallback if no file uploaded)
sample_text = """
Harry Potter looked at the letter in disbelief. The owl had arrived just as he was finishing breakfast.
Hogwarts School of Witchcraft and Wizardry - he was finally going to learn magic properly.
Hermione Granger was already reading through all her textbooks, while Ron Weasley complained about the homework.
The Great Hall was filled with students talking excitedly about the new term.
Dumbledore stood at the podium, his twinkling eyes surveying the crowd.
Magic was everywhere - in the floating candles, in the enchanted ceiling showing the night sky,
in the very air they breathed. Voldemort's threat loomed in the distance, but for now,
they were safe within the castle walls.
"""

# Try to load uploaded file
uploaded = files.upload()
text_data = ""

if uploaded:
    for filename in uploaded.keys():
        print(f'\nUploaded file: {filename}')
        # Read the file
        if filename.endswith('.txt'):
            text_data = uploaded[filename].decode('utf-8')
        elif filename.endswith('.csv'):
            df = pd.read_csv(io.StringIO(uploaded[filename].decode('utf-8')), delimiter=';')
            # Assuming text is in a column named 'text' or 'content'
            text_columns = [col for col in df.columns if 'text' in col.lower() or 'content' in col.lower()]
            if text_columns:
                text_data = ' '.join(df[text_columns[0]].astype(str).tolist())
            else:
                text_data = ' '.join(df.iloc[:, 0].astype(str).tolist())
else:
    print("No file uploaded. Using sample Harry Potter text.")
    text_data = sample_text

print(f"\nDataset loaded successfully!")
print(f"Total characters: {len(text_data):,}")
print(f"Preview (first 500 chars):\n{text_data[:500]}...")

# ==================================================
# 2. PREPROCESSING
# ==================================================
print("\n" + "="*50)
print("2. PREPROCESSING")
print("="*50)

def clean_text(text):
    """
    Clean the text by removing unwanted characters and formatting
    """
    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)

    # Remove special characters but keep basic punctuation
    text = re.sub(r'[^\w\s.,!?\'"-]', '', text)

    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text)

    # Remove extra newlines
    text = text.replace('\n', ' ').replace('\r', ' ')

    # Trim spaces
    text = text.strip()

    return text

# Clean the text
cleaned_text = clean_text(text_data)
print(f"Text cleaned successfully!")

# Tokenize using NLTK
print("\nTokenizing text...")
tokens = word_tokenize(cleaned_text)
print(f"Total tokens: {len(tokens):,}")
print(f"Sample tokens: {tokens[:20]}")

# Create sequences of tokens
print("\nCreating sequences...")
sequence_length = 10  # Number of words to use as context
sequences = []

for i in range(sequence_length, len(tokens)):
    seq = tokens[i-sequence_length:i+1]
    sequences.append(' '.join(seq))

print(f"Total sequences created: {len(sequences):,}")
print(f"Sample sequence: {sequences[0]}")

# ==================================================
# 3. TOKENIZATION & SEQUENCE PREPARATION
# ==================================================
print("\n" + "="*50)
print("3. TOKENIZATION & SEQUENCE PREPARATION")
print("="*50)

# Initialize tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sequences)

# Convert text to sequences
sequences = tokenizer.texts_to_sequences(sequences)

# Vocabulary size
vocab_size = len(tokenizer.word_index) + 1
print(f"Vocabulary size: {vocab_size}")

# Convert to numpy array
sequences = np.array(sequences)

# Split into predictors (X) and target (y)
X = sequences[:, :-1]  # All words except last
y = sequences[:, -1]   # Last word

# One-hot encode the target
y = tf.keras.utils.to_categorical(y, num_classes=vocab_size)

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")

# Pad sequences (though they're already the same length)
X = pad_sequences(X, maxlen=sequence_length, padding='pre')

# ==================================================
# 4. MODEL BUILDING
# ==================================================
print("\n" + "="*50)
print("4. MODEL BUILDING")
print("="*50)

# Model parameters
embedding_dim = 100
lstm_units = 256

# Build the model
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=sequence_length),
    LSTM(lstm_units, return_sequences=True),
    Dropout(0.3),
    LSTM(lstm_units // 2),
    Dropout(0.3),
    Dense(vocab_size // 2, activation='relu'),
    Dense(vocab_size, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Model Summary:")
model.summary()

# ==================================================
# 5. TRAINING
# ==================================================
print("\n" + "="*50)
print("5. TRAINING")
print("="*50)

# Callbacks
callbacks = [
    EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    ModelCheckpoint('best_model.h5', monitor='val_loss', save_best_only=True)
]

# Train-validation split
split_idx = int(0.8 * len(X))
X_train, X_val = X[:split_idx], X[split_idx:]
y_train, y_val = y[:split_idx], y[split_idx:]

print(f"Training samples: {len(X_train):,}")
print(f"Validation samples: {len(X_val):,}")

# Train the model
epochs = 20
batch_size = 64

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=epochs,
    batch_size=batch_size,
    callbacks=callbacks,
    verbose=1
)

print("\nTraining completed!")

# ==================================================
# 6. TEXT GENERATION
# ==================================================
print("\n" + "="*50)
print("6. TEXT GENERATION")
print("="*50)

def generate_text(seed_text, num_words=100, temperature=1.0):
    """
    Generate text given a seed text
    """
    generated_text = seed_text

    for _ in range(num_words):
        # Tokenize the seed text
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences([token_list], maxlen=sequence_length, padding='pre')

        # Predict the next word
        predictions = model.predict(token_list, verbose=0)[0]

        # Apply temperature
        predictions = np.log(predictions) / temperature
        exp_preds = np.exp(predictions)
        predictions = exp_preds / np.sum(exp_preds)

        # Sample from the distribution
        probas = np.random.multinomial(1, predictions, 1)

        # Get the index of the predicted word
        predicted_index = np.argmax(probas)

        # Convert index to word
        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted_index:
                output_word = word
                break

        # Add the word to the seed text and generated text
        seed_text += " " + output_word
        generated_text += " " + output_word

    return generated_text

# Test with different seed texts
print("\nGenerating text with different seed texts...")
print("-" * 50)

# Seed text 1
seed1 = "harry potter looked at the"
print(f"\nSeed Text 1: '{seed1}'")
generated1 = generate_text(seed1, num_words=50)
print(f"Generated Text 1:\n{generated1}")

# Seed text 2
seed2 = "hermione opened her book and"
print(f"\nSeed Text 2: '{seed2}'")
generated2 = generate_text(seed2, num_words=75)
print(f"Generated Text 2:\n{generated2}")

# Seed text 3
seed3 = "in the great hall dumbledore"
print(f"\nSeed Text 3: '{seed3}'")
generated3 = generate_text(seed3, num_words=100)
print(f"Generated Text 3:\n{generated3}")

# ==================================================
# 7. EVALUATION & VISUALIZATION
# ==================================================
print("\n" + "="*50)
print("7. EVALUATION & VISUALIZATION")
print("="*50)

import matplotlib.pyplot as plt

# Plot training history
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Loss plot
ax1.plot(history.history['loss'], label='Training Loss')
ax1.plot(history.history['val_loss'], label='Validation Loss')
ax1.set_title('Model Loss')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()
ax1.grid(True)

# Accuracy plot
ax2.plot(history.history['accuracy'], label='Training Accuracy')
ax2.plot(history.history['val_accuracy'], label='Validation Accuracy')
ax2.set_title('Model Accuracy')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy')
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()

# ==================================================
# 8. SAVE MODEL & TOKENIZER
# ==================================================
print("\n" + "="*50)
print("8. SAVING MODEL & TOKENIZER")
print("="*50)

# Save the model
model.save('harry_potter_text_generator.h5')
print("✓ Model saved as 'harry_potter_text_generator.h5'")

# Save the tokenizer
import pickle
with open('tokenizer.pkl', 'wb') as f:
    pickle.dump(tokenizer, f)
print("✓ Tokenizer saved as 'tokenizer.pkl'")

# ==================================================
# 9. IMPROVEMENTS & CHALLENGES
# ==================================================
print("\n" + "="*50)
print("9. IMPROVEMENTS & CHALLENGES")
print("="*50)

print("""
CHALLENGES FACED:
1. Limited dataset size for comprehensive training
2. Vocabulary coverage for rare magical terms
3. Maintaining consistent narrative style
4. Balancing creativity vs coherence

IMPROVEMENTS TO TRY:
1. Use larger dataset with more Harry Potter content
2. Implement attention mechanism
3. Use GPT-2 or Transformer architecture
4. Add beam search for better text generation
5. Implement temperature scheduling
6. Use character-level tokens for rare words
7. Add more LSTM layers or increase hidden units
8. Implement bidirectional LSTM
9. Use pretrained word embeddings
10. Add more diverse training data from fantasy genre
""")

print("\n" + "="*50)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("="*50)
```
